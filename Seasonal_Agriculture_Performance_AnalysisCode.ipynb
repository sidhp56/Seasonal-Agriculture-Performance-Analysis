## 7. Data Cleaning and Preparation

For numeric variables, missing observations are replaced using the **median**, which is less affected by extreme values than the mean.

Categorical variables, if missing, are replaced with the **mode**.

```python
# Identify numeric and categorical columns
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(include="object").columns.tolist()

# Fill numeric missing values with median
for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values with mode
for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after cleaning:")
print(df.isnull().sum().sum())
```

# 31. Statistical Test: Does Yield Differ Across Seasons?

A one-way ANOVA is used to test whether mean yield differs significantly among the seasons.

- **H0:** Mean yield is the same across seasons.
- **H1:** At least one season has a different mean yield.
- Significance level: 0.05

```python
season_groups = [
    group["Yield_Tonnes_Ha"].values
    for _, group in df.groupby("Season")
]

anova_stat, anova_p = stats.f_oneway(*season_groups)

print("ANOVA F-statistic:", round(anova_stat, 4))
print("ANOVA p-value:", anova_p)

if anova_p < 0.05:
    print("Result: Reject H0. Yield differs significantly across at least some seasons.")
else:
    print("Result: Fail to reject H0. No statistically significant seasonal difference in mean yield was detected.")
```

# 32. Statistical Test: Does Profit Differ Across Seasons?

This ANOVA checks whether average profit differs among seasons.

```python
profit_groups = [
    group["Profit_INR"].values
    for _, group in df.groupby("Season")
]

profit_anova_stat, profit_anova_p = stats.f_oneway(*profit_groups)

print("ANOVA F-statistic:", round(profit_anova_stat, 4))
print("ANOVA p-value:", profit_anova_p)

if profit_anova_p < 0.05:
    print("Result: Reject H0. Profit differs significantly across at least some seasons.")
else:
    print("Result: Fail to reject H0. No statistically significant seasonal difference in mean profit was detected.")
```

# 35. Outlier Detection for Profit

The IQR method is used to identify unusually high or low profit observations. These observations should be investigated rather than automatically deleted.

```python
q1 = df["Profit_INR"].quantile(0.25)
q3 = df["Profit_INR"].quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

profit_outliers = df[(df["Profit_INR"] < lower) | (df["Profit_INR"] > upper)]

print("Lower threshold:", lower)
print("Upper threshold:", upper)
print("Number of profit outliers:", len(profit_outliers))

display(profit_outliers.head(10))
```
